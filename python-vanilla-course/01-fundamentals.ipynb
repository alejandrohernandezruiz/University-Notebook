## 1. Object References and Mutability

Python variables are names that refer to objects in memory. Assigning one variable to another does not necessarily create a new object.

- **Immutable objects:** `int`, `float`, `bool`, `str`, and `tuple`. Their contents cannot be changed after creation.
- **Mutable objects:** `list`, `dict`, and `set`. Their contents can be changed in place.

This distinction is important when assigning variables or passing objects to functions.

### 1.1 Immutable objects

When an immutable value is reassigned or updated, the variable refers to a new object. The original value remains unchanged.

```python
a: int = 10
b = a

b += 5

print(f"a = {a}")
print(f"b = {b}")
```

### 1.2 Mutable objects

When two variables refer to the same mutable object, a change made through either variable is visible through both names.

```python
list_1: list[int] = [1, 2, 3]
list_2 = list_1

list_2.append(4)

print(f"list_1 = {list_1}")
print(f"list_2 = {list_2}")
```

### 1.3 Copying a list

Use `.copy()` or slicing (`[:]`) to create a shallow copy of a list. The new list has its own outer structure, so adding or removing elements does not affect the original list.

For nested lists or other nested mutable objects, a shallow copy still shares the inner objects.

```python
list_3 = list_1.copy()
list_3.append(5)

print(f"Original: {list_1}")
print(f"Copy:     {list_3}")
```

### 1.4 Mutability and functions

A function that receives a mutable object can modify that object. Those changes are visible outside the function because the caller and the function refer to the same object.

```python
def add_item(items: list[int]) -> None:
    items.append(99)


values = [1, 2, 3]
add_item(values)

print(values)
```

## 2. Read-Only NumPy Arrays

NumPy arrays are mutable by default. However, an array can be marked as read-only by setting its `flags.writeable` attribute to `False`.

An attempt to assign a new value through that array will raise a `ValueError`.

NumPy scalar values, such as `np.int64` and `np.float64`, are immutable objects.

```python
import numpy as np

arr_read_only = np.array([10, 20, 30])
arr_read_only.flags.writeable = False

print(arr_read_only)
print(f"Writeable: {arr_read_only.flags.writeable}")
```

## 3. Modern Control Flow

Python provides concise expressions for simple conditional assignments. These expressions complement traditional `if` statements; they do not replace them when several statements or branches are required.

### 3.1 Conditional expression (ternary operator)

A conditional expression selects one of two values based on a condition.

**Syntax:** `value_if_true if condition else value_if_false`

```python
pressure: float = 104.2

if pressure > 100.0:
    status = "CRITICAL"
else:
    status = "NORMAL"

print(status)
```

```python
status = "CRITICAL" if pressure > 100.0 else "NORMAL"

print(status)
```

Use a conditional expression for a simple value assignment. Prefer a regular `if` / `elif` / `else` block when the logic becomes more complex.

### 3.2 The Walrus Operator (`:=`)

The Walrus Operator, introduced in Python 3.8, assigns a value to a variable as part of an expression.

It can be useful when a value is needed both for a condition and inside the corresponding block.

```python
readings = {"name": "Ana", "age": 20}

n = len(readings)

if n > 3:
    print(f"Valid sample with {n} active sensors.")
else:
    print(f"Only {n} entries found.")
```

```python
if (n := len(readings)) > 3:
    print(f"Valid sample with {n} active sensors.")
else:
    print(f"Only {n} entries found.")
```

## 4. Pythonic Iteration

Python's built-in iteration tools often make loops clearer and less error-prone than manually managing indexes.

- `enumerate()` provides an index and its corresponding value.
- `zip()` combines values from multiple iterables and stops when the shortest iterable is exhausted.

### 4.1 Using `enumerate()`

Use `enumerate()` when you need both the position and the value of each item. The optional `start` argument controls the first index.

```python
temperatures: list[float] = [22.5, 24.0, 19.8]

for index, temperature in enumerate(temperatures, start=1):
    print(f"Reading #{index}: {temperature} °C")
```

### 4.2 Using `zip()`

Use `zip()` to process corresponding elements from multiple iterables without manually accessing their indexes.

```python
sensors: list[str] = ["S1", "S2", "S3"]
temperatures: list[float] = [22.5, 24.0, 19.8]

for sensor, temperature in zip(sensors, temperatures):
    print(f"Sensor {sensor} -> {temperature} °C")
```

## 5. Vanilla Python vs. NumPy: Vectorization

Standard Python loops process values one at a time. NumPy supports **vectorized operations**, allowing an operation to be applied to an entire array without writing an explicit Python loop.

For numerical workloads involving large, homogeneous arrays, NumPy operations are often substantially faster because much of the work is performed by optimized compiled code.

Example: converting temperatures from Celsius to Kelvin.

\[
T_K = T_C + 273.15
\]

### 5.1 Vanilla Python

With a standard Python list, create a new list and append each converted value inside a loop.

```python
temperatures_c: list[float] = [20.0, 25.5, 30.2, 18.0]
temperatures_k: list[float] = []

for temperature in temperatures_c:
    temperatures_k.append(temperature + 273.15)

print(temperatures_k)
```

### 5.2 NumPy vectorization

With a NumPy array, add the conversion offset directly to the entire array. NumPy applies the operation element by element.

```python
temperatures_c_array = np.array([20.0, 25.5, 30.2, 18.0])
temperatures_k_array = temperatures_c_array + 273.15

print(temperatures_k_array)
```
